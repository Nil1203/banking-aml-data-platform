```python
from pyspark.sql.types import *

audit_schema = StructType([
    StructField("job_name", StringType(), True),
    StructField("run_timestamp", TimestampType(), True),
    StructField("records_read", LongType(), True),
    StructField("records_written", LongType(), True),
    StructField("status", StringType(), True)
])

empty_df = spark.createDataFrame([], audit_schema)

empty_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "banking_catalog.banking_schema.audit_log"
    )
```

```python
spark.read.table(
    "banking_catalog.banking_schema.audit_log"
).printSchema()
```

```python
from datetime import datetime

audit_data = [
(
    "gold_pattern_summary",
    datetime.now(),
    370,
    8,
    "SUCCESS"
)
]

audit_df = spark.createDataFrame(
    audit_data,
    ["job_name",
     "run_timestamp",
     "records_read",
     "records_written",
     "status"]
)
```

```python
audit_df.write \
    .format("delta") \
    .mode("append") \
    .saveAsTable(
        "banking_catalog.banking_schema.audit_log"
    )
```
